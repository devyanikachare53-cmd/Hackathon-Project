```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("./heart.csv")
```

```python
df.head(10)
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
print("Count of duplicate rows : ", df.duplicated().sum())
```

```python
count_chol = (df['Cholesterol'] == 0).sum()
count_bp = (df['RestingBP'] == 0).sum()
print("Count of rows have Cholesterol zero : ", count_chol)
print("Count of rows have RestingBP zero : ", count_bp)
```

```python
df.describe()
```

```python
chol = df[df['Cholesterol'] != 0]
mean_chol = chol['Cholesterol'].mean()
df['Cholesterol'] = df['Cholesterol'].replace(0, round(mean_chol, 2))

bp = df[df['RestingBP'] != 0]
mean_bp = bp['RestingBP'].mean()
df['RestingBP'] = df['RestingBP'].replace(0, round(mean_bp, 2))
```

```python
df.describe()
```

```python
def plothistograms():
  plt.subplot(2, 2, 1)
  sns.histplot(df['Age'])

  plt.subplot(2, 2, 2)
  sns.histplot(df['RestingBP'])

  plt.subplot(2, 2, 3)
  sns.histplot(df['Cholesterol'])

  plt.subplot(2, 2, 4)
  sns.histplot(df['MaxHR'])

  plt.tight_layout()

plothistograms()
```

```python
print("Categorical Columns : ")
for col in df.columns:
  if df[col].dtype == 'object':
    print(df[col])
```

```python
print("Numerical Columns : ")
for col in df.columns:
  if df[col].dtype in ['int64', 'float64']:
    print(df[col])
```

```python
df_encoded = pd.get_dummies(df, columns= ['Sex', 'ChestPainType', 'RestingECG','ExerciseAngina', 'ST_Slope'], dtype = 'int64')

df_encoded
```

```python
print("Shape: ")
df_encoded.shape
```

```python
df_encoded.head(5)
```

```python
print("Final Shape: ")
df_encoded.shape
```

```python
print("Column Names : ")
df_encoded.columns
```

```python
from sklearn.preprocessing import StandardScaler

numeric_cols = ["Age",	"RestingBP",	"Cholesterol",	"FastingBS",	"MaxHR",	"Oldpeak"]
scaler = StandardScaler()
df_encoded[numeric_cols] = scaler.fit_transform(df_encoded[numeric_cols])
df_encoded
```

```python
X = df_encoded.drop('HeartDisease', axis = 1)
y = df_encoded['HeartDisease']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state = 42)
```

## Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print("Predicted Value : ", y_pred[:11])
print("Actual Value : ", y_test[:11].values)
```

```python
from sklearn.metrics import confusion_matrix

conf_mat = confusion_matrix(y_test, y_pred)
print("Confusion Matrix : ")
print(conf_mat)

print("\nTrue Negative TN : ", conf_mat[0, 0])
print("False Positive FP : ", conf_mat[0, 1])
print("False Negative FN : ", conf_mat[1, 0])
print("True Positive TP : ", conf_mat[1, 1])
```

```python
from sklearn.metrics import classification_report

repo = classification_report(y_test, y_pred)
print("Classification Report : ")
print(repo)
```

```python
from sklearn.metrics import accuracy_score

print("Accuracy Score : ", accuracy_score(y_test, y_pred))
```

## Decision Tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier

model_DT = DecisionTreeClassifier()
model_DT.fit(X_train, y_train)
```

```python
y_pred_DT = model_DT.predict(X_test)

print("Predicted Value : ", y_pred_DT[:11])
print("Actual Value : ", y_test[:11].values)
```

```python
from sklearn.metrics import confusion_matrix

conf_mat = confusion_matrix(y_test, y_pred_DT)
print("Confusion Matrix : ")
print(conf_mat)

print("\nTrue Negative TN : ", conf_mat[0, 0])
print("False Positive FP : ", conf_mat[0, 1])
print("False Negative FN : ", conf_mat[1, 0])
print("True Positive TP : ", conf_mat[1, 1])
```

```python
from sklearn.metrics import accuracy_score

print("Accuracy Score : ", accuracy_score(y_test, y_pred_DT))
```
